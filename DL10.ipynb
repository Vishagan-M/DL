```python
# Download the pretrained FFHQ model pickle file
!wget -O ffhq.pkl https://nvlabs-fi-cdn.nvidia.com/stylegan2-ada-pytorch/pretrained/ffhq.pkl
```

```python
import sys
import os

# Ensure the cloned repository is in the system path
sys.path.insert(0, os.path.abspath('./stylegan2-ada-pytorch'))

import torch
from torchvision.utils import save_image
import legacy

# Load pretrained StyleGAN2
device = "cuda" if torch.cuda.is_available() else "cpu"
G = legacy.load_network_pkl(
    open("ffhq.pkl","rb"))["G_ema"].to(device)

# Generate image
os.makedirs("generated", exist_ok=True)

z = torch.randn(1, G.z_dim).to(device)
label = torch.zeros(1, G.c_dim).to(device)

with torch.no_grad():
    img = G(z, label, truncation_psi=0.5)
    img = (img + 1) / 2

    save_image(img, "generated/image.png")

    print("Image generated successfully")
```
