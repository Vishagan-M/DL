```python
import tensorflow as tf
import numpy as np

# Load text
path = tf.keras.utils.get_file(
    "shakespeare.txt",
    "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt")

text = open(path).read().lower()

chars = sorted(set(text))
c2i = {c:i for i,c in enumerate(chars)}
i2c = {i:c for i,c in enumerate(chars)}

# Sequences
n = 40
X = np.array([[c2i[c] for c in text[i:i+n]]
              for i in range(len(text)-n)])
y = np.array([c2i[text[i+n]] for i in range(len(text)-n)])

# LSTM
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(len(chars),64),
    tf.keras.layers.LSTM(128),
    tf.keras.layers.Dense(len(chars),activation="softmax")
])

model.compile(loss="sparse_categorical_crossentropy",
              optimizer="adam")
model.fit(X,y,batch_size=128,epochs=1)

# Generate text
seed = "shall i compare thee"
seq = [c2i[c] for c in seed]

for _ in range(100):
    p = model.predict(np.array(seq[-n:]).reshape(1,-1),verbose=0)[0]
    seq.append(np.random.choice(len(chars),p=p))

print("Generated Text:")
print(''.join(i2c[i] for i in seq))
```
